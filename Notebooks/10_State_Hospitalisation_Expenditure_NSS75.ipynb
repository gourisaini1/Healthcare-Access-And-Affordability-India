# State-level hospitalisation medical expenditure (NSS 75th round)

This notebook analyses average medical expenditure during hospital stay per hospitalisation case, excluding childbirth, from NSS 75th Round (July 2017–June 2018), Table A17. It retains rural, urban, public, private and all-hospital estimates.

**Important:** Table A17 reports gross medical expenditure incurred during hospital stay, not net out-of-pocket expenditure after reimbursements. State-level net OOPE is not claimed. The 2017–18 state/UT boundaries also predate the creation of Ladakh and the merger of Dadra & Nagar Haveli with Daman & Diu.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DATA = Path('../Dataset')
OUT = Path('../Visualisations')
df = pd.read_csv(DATA / 'state_hospitalisation_medical_expenditure_nss75.csv')
assert len(df) == 36 and df.state_ut_2017.is_unique
assert df.reference_period.eq('2017-18').all()
assert df.all_hospitals_rural_urban_inr.gt(0).all()
print(f'State/UT observations: {len(df)}')
print(f'All-hospital rural+urban mean: ₹{df.all_hospitals_rural_urban_inr.mean():,.0f}')
print(f'All-hospital rural+urban median: ₹{df.all_hospitals_rural_urban_inr.median():,.0f}')
print('Lowest:', df.loc[df.all_hospitals_rural_urban_inr.idxmin(), 'state_ut_2017'], int(df.all_hospitals_rural_urban_inr.min()))
print('Highest:', df.loc[df.all_hospitals_rural_urban_inr.idxmax(), 'state_ut_2017'], int(df.all_hospitals_rural_urban_inr.max()))
```

## Interpretation boundary

The all-hospitals rural+urban column is used for the headline comparison. State differences may reflect hospital type, case mix, treatment intensity, prices and access patterns as well as affordability. These averages do not measure household capacity to pay, reimbursement, insurance claim settlement or catastrophic spending. They should not be interpreted as net OOPE.

```python
plot_df = df.sort_values('all_hospitals_rural_urban_inr', ascending=True)
fig, ax = plt.subplots(figsize=(10, 10))
ax.barh(plot_df.state_ut_2017, plot_df.all_hospitals_rural_urban_inr)
ax.set_xlabel('Average medical expenditure per hospitalisation case (₹)')
ax.set_title('Average hospital-stay medical expenditure by state/UT, NSS 75th round (2017–18)')
ax.grid(axis='x', alpha=0.25)
fig.tight_layout()
fig.savefig(OUT / '13_nss75_hospital_medical_expenditure_by_state.svg', format='svg', bbox_inches='tight')
plt.show()
```
