# State-wise government hospitals and beds

This notebook explores state/UT counts of government hospitals and beds reproduced in RBI Handbook of Statistics on Indian States 2024–25, Table 17. The RBI table does not state the distinct observation year; the edition year is not treated as the observation year.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(Path('../Dataset/state_government_hospitals_and_beds_rbi_handbook_2024_25.csv'))
df.head()
```

## 1. Validate data and source totals

```python
print('State/UT rows:', len(df))
print('Duplicate names:', df['state_ut'].duplicated().sum())
print('Missing values:\n', df.isna().sum())
assert df['state_ut'].is_unique
print('Sum of state/UT beds:', df['beds_total'].sum())
print('RBI all-India beds total (includes any centrally administered coverage): 840455')
```

## 2. Government hospital beds by state/UT

Absolute counts are shown. Population-adjusted bed availability would be more suitable for comparing access, but requires a compatible population denominator and reference year.

```python
ranked = df.dropna(subset=['beds_total']).sort_values('beds_total')
plt.figure(figsize=(10, 11))
plt.barh(ranked['state_ut'], ranked['beds_total'])
plt.xlabel('Government hospital beds (count)')
plt.ylabel('State / UT')
plt.title('Government hospital beds by state/UT')
plt.tight_layout()
plt.show()
```

## 3. Government hospitals by state/UT

```python
ranked_h = df.dropna(subset=['government_hospitals_total']).sort_values('government_hospitals_total')
plt.figure(figsize=(10, 11))
plt.barh(ranked_h['state_ut'], ranked_h['government_hospitals_total'])
plt.xlabel('Government hospitals (count)')
plt.ylabel('State / UT')
plt.title('Government hospitals by state/UT')
plt.tight_layout()
plt.show()
```

## Interpretation cautions

- The RBI table does not specify a distinct observation year; do not label these counts as 2024–25 observations.
- Absolute counts favour more populous states and should not be used alone to rank access.
- Population-adjusted beds per 10,000 people, facility distribution, staffing and utilisation would improve the analysis.
- A dash in the source is unavailable data, not zero.
