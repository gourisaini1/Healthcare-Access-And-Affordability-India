# Bivariate correlation and simple linear regression: public health expenditure and infant mortality

This notebook estimates an unadjusted, cross-sectional association between the 2024–25 Revised Estimate of health expenditure as a share of aggregate expenditure and 2023 infant mortality rates. It fits a simple bivariate line only; it does not include socioeconomic controls. The years differ, the spending measure is a budget share rather than a per-capita input, and no causal inference is attempted.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr, linregress

df = pd.read_csv(Path('../Dataset/state_health_spending_imr_matched_exploratory.csv'))
df.head()
```

## 1. Data checks

```python
assert df['state_ut'].is_unique
assert len(df) == 31
assert df['health_expenditure_share_2024_25_re_pct'].between(0, 100).all()
assert df['infant_mortality_rate_2023_per_1000_live_births'].ge(0).all()
print(df.describe())
```

## 2. Scatter plot

```python
x = df['health_expenditure_share_2024_25_re_pct']
y = df['infant_mortality_rate_2023_per_1000_live_births']
plt.figure(figsize=(9, 6))
plt.scatter(x, y)
for _, row in df.iterrows():
    plt.annotate(row['state_ut'], (row['health_expenditure_share_2024_25_re_pct'], row['infant_mortality_rate_2023_per_1000_live_births']), fontsize=7, alpha=0.8)
plt.xlabel('Health expenditure (% of aggregate expenditure), 2024–25 RE')
plt.ylabel('Infant mortality rate, 2023 (per 1,000 live births)')
plt.title('State health expenditure share and infant mortality')
plt.tight_layout()
plt.show()
```

## 3. Correlation and simple linear fit

```python
pearson_r, pearson_p = pearsonr(x, y)
spearman_rho, spearman_p = spearmanr(x, y)
fit = linregress(x, y)
print(f'Pearson correlation: {pearson_r:.3f}; p-value: {pearson_p:.3f}')
print(f'Spearman correlation: {spearman_rho:.3f}; p-value: {spearman_p:.3f}')
print(f'Slope: {fit.slope:.3f} IMR points per 1 percentage-point budget share')
print(f'R-squared: {fit.rvalue**2:.3f}')
print(f'Observations: {len(df)}')
```

## 4. Interpretation

The Pearson correlation in this matched sample is approximately 0.044, indicating almost no linear cross-sectional association between these two particular measures. This is not evidence that public spending is ineffective. The expenditure variable is a share of aggregate state expenditure, not the amount spent per resident or the resources actually reaching frontline services. IMR is for calendar year 2023, whereas spending is a 2024–25 Revised Estimate. The sample is small and other determinants—income, maternal education, nutrition, sanitation, fertility, urbanisation, health workforce, and data quality—are not controlled for.

The simple regression is descriptive only. It must not be described as a causal estimate or a policy impact evaluation.

## 5. Suggested next improvements

- Use health expenditure per capita or as a percentage of GSDP with matched year definitions.
- Add population-adjusted beds and health workforce availability.
- Add controls for income, literacy, sanitation, nutrition and urbanisation.
- Prefer multiple years of state-level panel data before attempting fixed-effects analysis.
