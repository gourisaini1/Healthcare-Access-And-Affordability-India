# 04 — State-level healthcare infrastructure

Input: a reviewed state/UT table transcribed from the Ministry of Health and Family Welfare's Rural Health Statistics 2021–22 report. Save it as `Dataset/Raw_Data/rhs_state_infrastructure.csv`.

Required: `state_ut` plus available numeric columns such as `sub_centres`, `phcs`, `chcs`, `doctors_required`, `doctors_in_position`, `specialists_required`, `specialists_in_position`. Preserve source definitions and rural/urban scope. Source PDF: https://www.mohfw.gov.in/sites/default/files/RHS%202021-22.pdf

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import pandas as pd
ROOT=Path.cwd()
if not (ROOT/"README.md").exists() and (ROOT.parent/"README.md").exists(): ROOT=ROOT.parent
RAW=ROOT/"Dataset"/"Raw_Data"; OUT=ROOT/"Dataset"/"Processed_Data"; FIG=ROOT/"Dataset"/"Notebooks"/"Visualisations"
RAW.mkdir(parents=True,exist_ok=True); OUT.mkdir(parents=True,exist_ok=True); FIG.mkdir(parents=True,exist_ok=True)
p=RAW/"rhs_state_infrastructure.csv"
if not p.exists(): raise FileNotFoundError("Add reviewed RHS state table at Dataset/Raw_Data/rhs_state_infrastructure.csv; see Dataset/Raw_Data/README.md.")
df=pd.read_csv(p); assert "state_ut" in df.columns
nums=df.select_dtypes(include="number").columns; display(df.head()); display(df[nums].describe().T)
for col in [c for c in ["sub_centres","phcs","chcs"] if c in df.columns]:
    g=df.sort_values(col,ascending=False); fig,ax=plt.subplots(figsize=(11,7)); sns.barplot(data=g,x=col,y="state_ut",ax=ax); ax.set(title=f"{col.replace('_',' ').title()} by State/UT",xlabel="Facility count",ylabel=""); fig.tight_layout(); fig.savefig(FIG/f"rhs_{col}_by_state.png",dpi=180,bbox_inches="tight"); plt.show()
```

Counts alone are not access rates. Adjust for population or use source-reported population coverage. State size and rural/urban definitions matter; do not compare unlike denominators.
