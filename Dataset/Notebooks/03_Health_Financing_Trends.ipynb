# 03 — Health financing trends

Plot national out-of-pocket expenditure and public financing indicators from the retrieved data.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import pandas as pd
ROOT=Path.cwd()
if not (ROOT/"README.md").exists() and (ROOT.parent/"README.md").exists(): ROOT=ROOT.parent
RAW=ROOT/"Dataset"/"Raw_Data"; OUT=ROOT/"Dataset"/"Processed_Data"; FIG=ROOT/"Dataset"/"Notebooks"/"Visualisations"
RAW.mkdir(parents=True,exist_ok=True); OUT.mkdir(parents=True,exist_ok=True); FIG.mkdir(parents=True,exist_ok=True)
p=OUT/"india_health_financing_clean.csv"
if not p.exists(): raise FileNotFoundError("Run notebooks 01 and 02 first.")
df=pd.read_csv(p); sns.set_theme(style="whitegrid")
labels={"oop_share":"Out-of-pocket expenditure (% of current health expenditure)","gov_share":"Government health expenditure (% of current health expenditure)","health_gdp":"Current health expenditure (% of GDP)"}
for key,title in labels.items():
    g=df[df.indicator==key].sort_values("year")
    if g.empty: print("No observations:",key); continue
    fig,ax=plt.subplots(figsize=(10,5)); ax.plot(g.year,g.value,marker="o",linewidth=2); ax.set(title=title,xlabel="Year",ylabel="Percent"); fig.tight_layout(); fig.savefig(FIG/f"{key}_trend.png",dpi=180,bbox_inches="tight"); plt.show()
latest=df.sort_values("year").groupby("indicator").tail(1)[["indicator","year","value"]]
display(latest); latest.to_csv(OUT/"latest_financing_indicators.csv",index=False)
```
